```python
import sys, glob
sys.path.insert(0, "/opt/spark/python")
sys.path.insert(0, glob.glob("/opt/spark/python/lib/py4j*.zip")[0])

from pyspark.sql import SparkSession, functions as F

spark = (SparkSession.builder
    .appName("Helpfulness")
    .master("local[4]")
    .config("spark.driver.memory", "5g")
    .config("spark.driver.maxResultSize", "1g")
    .config("spark.sql.shuffle.partitions", "16")
    .config("spark.hadoop.dfs.replication", "1")
    .getOrCreate())

HDFS = "hdfs://namenode:9000"
```

```python
from pyspark.sql import Window

CUTOFF = "2016-12-31"     # ubah di sini kalau mau uji sensitivitas

v2 = spark.read.parquet(f"{HDFS}/data/ml/features_v2")
day = (F.datediff(F.lit("2018-10-02").cast("date"), F.lit("1970-01-01").cast("date"))
       - F.col("age_days"))
v3 = v2.withColumn("day", day).withColumn("tgl", F.date_sub(F.lit("2018-10-02").cast("date"), F.col("age_days")))

w_prod_past = Window.partitionBy("asin").orderBy("day").rangeBetween(Window.unboundedPreceding, -1)

v3 = (v3
  .withColumn("tahun", F.year("tgl"))
  .withColumn("prod_prev_avg", F.avg("overall").over(w_prod_past))      # null = ulasan pertama
  .withColumn("rating_dev_past", F.when(F.col("prod_prev_avg").isNotNull(),
                                        F.col("overall") - F.col("prod_prev_avg")))
  .withColumn("abs_rating_dev_past", F.abs(F.col("rating_dev_past")))
  .drop("day"))

v3.write.mode("overwrite").parquet(f"{HDFS}/data/ml/features_v3")

v3 = spark.read.parquet(f"{HDFS}/data/ml/features_v3")
lab = v3.filter(F.col("tgl") <= F.lit(CUTOFF).cast("date"))
inf = v3.filter(F.col("tgl") >  F.lit(CUTOFF).cast("date"))

b = F.pmod(F.hash("reviewerID"), F.lit(20))        # 0-13 train, 14-16 validasi, 17-19 test
lab.filter(b < 14).write.mode("overwrite").parquet(f"{HDFS}/data/ml/train_v3")
lab.filter((b >= 14) & (b < 17)).write.mode("overwrite").parquet(f"{HDFS}/data/ml/val_v3")
lab.filter(b >= 17).write.mode("overwrite").parquet(f"{HDFS}/data/ml/test_v3")
inf.write.mode("overwrite").parquet(f"{HDFS}/data/ml/infer_v3")
```

```python
for nm in ["train_v3", "val_v3", "test_v3", "infer_v3"]:
    d = spark.read.parquet(f"{HDFS}/data/ml/{nm}")
    n = d.count()
    print(f"{nm:9s} baris={n:>9,}  helpful={d.agg(F.avg('label')).first()[0]:.3%}")
```

```python
import time
from pyspark.ml import Pipeline
from pyspark.ml.feature import StringIndexer, Imputer, VectorAssembler
from pyspark.ml.classification import GBTClassifier
from pyspark.ml.evaluation import BinaryClassificationEvaluator, MulticlassClassificationEvaluator

FILL = {"brand": "Other", "main_cat": "Unknown", "user_prev_help": -1.0, "prod_avg_loo": -1.0,
        "rating_dev": 0.0, "abs_rating_dev": 0.0, "prod_prev_avg": -1.0,
        "rating_dev_past": 0.0, "abs_rating_dev_past": 0.0}

tr = spark.read.parquet(f"{HDFS}/data/ml/train_v3").fillna(FILL)
va = spark.read.parquet(f"{HDFS}/data/ml/val_v3").fillna(FILL)

n = tr.count(); n_pos = tr.filter("label = 1").count(); w_pos = (n - n_pos) / n_pos
tr = tr.withColumn("w", F.when(F.col("label") == 1, F.lit(w_pos)).otherwise(F.lit(1.0))).cache()
va = va.cache(); va.count()
print(f"bobot kelas helpful: {w_pos:.2f}")

def evaluate(pred):
    out = {"PR-AUC": BinaryClassificationEvaluator(labelCol="label", metricName="areaUnderPR").evaluate(pred),
           "ROC-AUC": BinaryClassificationEvaluator(labelCol="label", metricName="areaUnderROC").evaluate(pred)}
    for m in ["precisionByLabel", "recallByLabel", "fMeasureByLabel"]:
        out[m] = MulticlassClassificationEvaluator(labelCol="label", metricName=m, metricLabel=1.0).evaluate(pred)
    return out

STATIK = ["overall", "verified", "text_len", "word_count", "has_price"]
KAUSAL = {
  "tahun":    ["tahun"],
  "posisi_k": ["prod_prev_n", "prod_gap_days"],
  "riwayat":  ["user_prev_n", "user_prev_help"],
  "teks":     ["summary_len", "n_exclaim", "n_question", "n_sentences", "has_url",
               "has_linebreak", "upper_ratio", "digit_ratio", "avg_word_len"],
  "deviasi_k":["prod_prev_avg", "rating_dev_past", "abs_rating_dev_past"],
}
SEMUA_K = STATIK + sum(KAUSAL.values(), [])
SEMUA_A = ["overall", "verified", "age_days", "text_len", "word_count", "has_price", "prod_avg_rating",
           "prod_review_count", "user_review_count", "summary_len", "n_exclaim", "n_question",
           "n_sentences", "has_url", "has_linebreak", "upper_ratio", "digit_ratio", "avg_word_len",
           "user_prev_n", "user_prev_help", "prod_prev_n", "rel_pos", "prod_gap_days",
           "prod_avg_loo", "rating_dev", "abs_rating_dev"]

results_b = {}
def run_b(tag, feats):
    t0 = time.time()
    stages = [Imputer(strategy="median", inputCols=["price_clean"], outputCols=["price_imp"]),
              StringIndexer(inputCols=["brand", "main_cat"], outputCols=["brand_idx", "main_cat_idx"],
                            handleInvalid="keep"),
              VectorAssembler(inputCols=feats + ["price_imp", "brand_idx", "main_cat_idx"], outputCol="features"),
              GBTClassifier(labelCol="label", featuresCol="features", weightCol="w",
                            maxIter=50, maxDepth=5, maxBins=64, seed=42)]
    model = Pipeline(stages=stages).fit(tr)
    fit_s = round(time.time() - t0)
    pred = model.transform(va).select("label", "prediction", "rawPrediction").cache()
    res = evaluate(pred); res["detik"] = fit_s; results_b[tag] = res
    pred.unpersist()
    print(tag, {k: round(v, 3) for k, v in res.items()})
    return model
```

```python
run_b("B0_statik", STATIK)
```

```python
run_b("B1_+tahun", STATIK + KAUSAL["tahun"])
```

```python
run_b("B2_+posisi_kausal", STATIK + KAUSAL["tahun"] + KAUSAL["posisi_k"])
```

```python
run_b("B3_+riwayat", STATIK + KAUSAL["tahun"] + KAUSAL["posisi_k"] + KAUSAL["riwayat"])
```

```python
run_b("B4_+teks", STATIK + KAUSAL["tahun"] + KAUSAL["posisi_k"] + KAUSAL["riwayat"] + KAUSAL["teks"])
mB = run_b("B5_semua_kausal", SEMUA_K)
```

```python
mB = run_b("B5_semua_kausal", SEMUA_K)
```

```python
run_b("A_semua_snapshot", SEMUA_A) 
```
